```python
!pip install torchmetrics Pillow requests

import os
import zipfile
import requests

# Download the UC Merced Satellite dataset
url = "https://hf.co/datasets/torchgeo/ucmerced/resolve/main/UCMerced_LandUse.zip"
zip_path = "UCMerced_LandUse.zip"

print("Downloading satellite dataset...")
r = requests.get(url, stream=True)
with open(zip_path, "wb") as f:
    for chunk in r.iter_content(chunk_size=8192):
        f.write(chunk)

print("Extracting images...")
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("dataset")

print("Dataset ready!")
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torchvision.transforms as T
from glob import glob

class SatelliteSRDataset(Dataset):
    def __init__(self, folder_path):
        # Search for all image files (.tif, .png, .jpg)
        self.image_paths = glob(os.path.join(folder_path, "**", "*.tif"), recursive=True) + \
                           glob(os.path.join(folder_path, "**", "*.png"), recursive=True)

        # Transformations: Blurry Input (64x64) vs Ground Truth Target (256x256)
        self.lr_transform = T.Compose([
            T.Resize((64, 64), interpolation=T.InterpolationMode.BICUBIC),
            T.ToTensor()
        ])
        self.hr_transform = T.Compose([
            T.Resize((256, 256), interpolation=T.InterpolationMode.BICUBIC),
            T.ToTensor()
        ])

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img = Image.open(self.image_paths[idx]).convert("RGB")
        lr_img = self.lr_transform(img) # Blur/Low-Res
        hr_img = self.hr_transform(img) # Crisp/High-Res
        return lr_img, hr_img

# Initialize DataLoader
dataset = SatelliteSRDataset("dataset")
train_loader = DataLoader(dataset, batch_size=16, shuffle=True)
print(f"Loaded {len(dataset)} satellite images.")
```

```python
import torch.nn as nn

class ResBlock(nn.Module):
    def __init__(self, channels=64):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(channels, channels, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, kernel_size=3, padding=1)
        )
    def forward(self, x):
        return x + self.conv(x) * 0.1

class EDSRModel(nn.Module):
    def __init__(self, scale_factor=4):
        super().__init__()
        self.head = nn.Conv2d(3, 64, kernel_size=3, padding=1)
        self.body = nn.Sequential(*[ResBlock(64) for _ in range(8)])
        self.body_conv = nn.Conv2d(64, 64, kernel_size=3, padding=1)

        # PixelShuffle upscales 64x64 input to 256x256 output
        self.upsample = nn.Sequential(
            nn.Conv2d(64, 64 * (scale_factor ** 2), kernel_size=3, padding=1),
            nn.PixelShuffle(scale_factor),
            nn.Conv2d(64, 3, kernel_size=3, padding=1)
        )

    def forward(self, x):
        x_head = self.head(x)
        x_body = self.body_conv(self.body(x_head)) + x_head
        return self.upsample(x_body)

# Push model to GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = EDSRModel(scale_factor=4).to(device)
print("Model initialized on:", device)
```

```python
import torch.optim as optim
from torchmetrics.image import PeakSignalNoiseRatio, StructuralSimilarityIndexMeasure

criterion = nn.L1Loss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# Performance Metrics
# Performance Metrics
psnr_metric = PeakSignalNoiseRatio(data_range=1.0).to(device)
ssim_metric = StructuralSimilarityIndexMeasure(data_range=1.0).to(device)
```

```python
epochs = 5 # Increase epochs (e.g., 20 or 50) for better quality

for epoch in range(1, epochs + 1):
    model.train()
    running_loss, total_psnr, total_ssim = 0.0, 0.0, 0.0

    for lr_imgs, hr_imgs in train_loader:
        lr_imgs, hr_imgs = lr_imgs.to(device), hr_imgs.to(device)

        optimizer.zero_grad()
        sr_imgs = model(lr_imgs) # AI generates enhanced image
        loss = criterion(sr_imgs, hr_imgs)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        total_psnr += psnr_metric(sr_imgs, hr_imgs).item()
        total_ssim += ssim_metric(sr_imgs, hr_imgs).item()

    avg_loss = running_loss / len(train_loader)
    avg_psnr = total_psnr / len(train_loader)
    avg_ssim = total_ssim / len(train_loader)

    print(f"Epoch [{epoch}/{epochs}] | Loss: {avg_loss:.4f} | PSNR: {avg_psnr:.2f} dB | SSIM: {avg_ssim:.4f}")
```
